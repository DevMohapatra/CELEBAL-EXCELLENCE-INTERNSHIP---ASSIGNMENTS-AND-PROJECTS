# Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output



````python
# TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
#  TOOL 3 (extra tool) Word Counter

def count_words(text: str) -> dict:
    """Count total words and unique words in a text."""
    try:
        words = text.split()
        return {"total_words": len(words), "unique_words": len(set(w.lower() for w in words))}
    except Exception:
        return {"total_words": 0, "unique_words": 0}
````

````python
#  TOOL 4 (extra tool): Text Case Converter

def convert_case(text: str, mode: str = "upper") -> str:
    """Convert text to uppercase or lowercase."""
    try:
        return text.upper() if mode == "upper" else text.lower() if mode == "lower" else text
    except Exception:
        return "Error in case conversion"
````

````python
#  LOGGING (Bonus section)

import datetime

query_log = []

def log_query(query: str, response: dict):
    query_log.append({
        "timestamp": datetime.datetime.now().isoformat(),
        "query": query,
        "type": response.get("type"),
        "result": response.get("result")
    })

def print_log():
    print(f"{'Timestamp':<28}{'Type':<14}{'Query'}")
    print("-" * 80)
    for entry in query_log:
        print(f"{entry['timestamp']:<28}{entry['type']:<14}{entry['query'][:40]}")
````

## Implement Agent Logic Below

Used conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
#  AGENT FUNCTION (TO IMPLEMENT)

def agent(query: str):
    query_lower = query.lower()

    try:
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "").strip()
            if not expression:
                response = {"type": "error", "result": "No expression found to calculate."}
            else:
                calc_result = calculator(expression)
                response = {"type": "error", "result": calc_result} if calc_result == "Error in calculation" else {"type": "calculation", "result": calc_result}

        elif "keywords" in query_lower:
            keywords_result = extract_keywords(query)
            response = {"type": "error", "result": "No keywords could be extracted."} if not keywords_result else {"type": "keywords", "result": keywords_result}

        elif "count words" in query_lower:
            response = {"type": "word_count", "result": count_words(query)}

        elif "uppercase" in query_lower:
            response = {"type": "case_conversion", "result": convert_case(query, "upper")}

        elif "lowercase" in query_lower:
            response = {"type": "case_conversion", "result": convert_case(query, "lower")}

        else:
            response = {"type": "general", "result": f"You asked: '{query}'. This is a general response since no specific tool matched."}

    except Exception as e:
        response = {"type": "error", "result": f"Unexpected error occurred: {str(e)}"}

    log_query(query, response)
    return response
````

##  Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
#  Test Cases

import json
queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query:", q)
    a=agent(q)
    print("Response:", json.dumps(a,indent=1))
    log_query(q,a)
    print_log()
    print("-" * 50)
````

````python
#  Interactive Mode

while True:
    user_input = input("Enter query (type 'exit' to stop): ")
    if user_input.lower() == "exit":
        break
    a=agent(user_input)
    print("Response:", json.dumps(a,indent=1))
    log_query(user_input,a)
    print_log()
````
